```python
import matplotlib.pyplot as plt
import numpy as np
fs=100
T=1

N=fs*T
t=np.arange(N)/fs

x=np.sin(2*np.pi*5*t)

plt.plot(t,x,marker=".")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
N=8
fs=8
freqs=np.fft.fftfreq(N,d=1/fs)  # d means the sampling time delay e 
freqs
```

Function,Primary Input → Output,Output Length (1D),Typical Use
fft(a),Complex or Real → Complex,N,General two-sided discrete Fourier transform
ifft(a),Complex → Complex or Real,N,Inverse of fft
rfft(a),Real → Complex,⌊N/2⌋+1,Efficient DFT for purely real signals (half the computation)
irfft(a),Complex → Real,N,Inverse of rfft back to time-domain real signal
"fftfreq(n, d)","Integers (N, sample spacing)",N,Generates two-sided frequency bins matching fft output
"rfftfreq(n, d)","Integers (N, sample spacing)",⌊N/2⌋+1,Generates one-sided frequency bins matching rfft output
fftshift(x),Any array,Same as input,Moves 0 Hz from the start to the center of the array
ifftshift(x),Any array,Same as input,Inverts fftshift (moves center back to index 0)
